The goal of this homework is to create a regression model for predicting the car fuel efficiency (column 'fuel_efficiency_mpg').

```python
import pandas as pd
import numpy as np

import seaborn as sns
from matplotlib import pyplot as plt
%matplotlib inline
```

```python
#1. load data

df = pd.read_csv('car_fuel_efficiency_2026.csv')
```

```python
df.fillna(0, inplace=True)
```

```python
df.describe().round()
```

```python
from sklearn.model_selection import train_test_split

df_full_train, df_test = train_test_split(df, test_size=0.2, random_state=1)
df_train, df_val = train_test_split(
    df_full_train, test_size=0.25, random_state=1
)
```

```python
df_train = df_train.reset_index(drop=True)
df_val = df_val.reset_index(drop=True)
df_test = df_test.reset_index(drop=True)
```

```python
y_train = (df_train.fuel_efficiency_mpg).values
y_val = (df_val.fuel_efficiency_mpg).values
y_test = (df_test.fuel_efficiency_mpg).values
```

```python
del df_train['fuel_efficiency_mpg']
del df_val['fuel_efficiency_mpg']
del df_test['fuel_efficiency_mpg']
```

```python
# Question 1. Which original feature is used for splitting the data? If you use DictVectorizer, map a one-hot feature such as origin=USA back to its original column name.

from sklearn.tree import DecisionTreeRegressor
from sklearn.feature_extraction import DictVectorizer
from sklearn.metrics import roc_auc_score
from sklearn.tree import export_text
```

```python
df_train.head()
```

```python
train_dicts = df_train.fillna(0).to_dict(orient='records')

dv = DictVectorizer(sparse=False)
X_train = dv.fit_transform(train_dicts)
```

```python
train_dicts
y_train
```

```python
dt = DecisionTreeRegressor(max_depth=1, random_state=1)
dt.fit(X_train, y_train)
```

```python
print(export_text(
    dt,
    feature_names=list(dv.get_feature_names_out())
))  
```

```python
feature_idx = dt.tree_.feature[0]
feature_name = dv.get_feature_names_out()[feature_idx]

print(feature_name)
```

```python
val_dicts = df_val.fillna(0).to_dict(orient='records')
X_val = dv.transform(val_dicts)

y_pred = dt.predict(X_val)
```

```python
# Question 2. Train a random forest regressor with these parameters:

from sklearn.ensemble import RandomForestRegressor
```

```python
rf = RandomForestRegressor(n_estimators=10,
                           n_jobs=-1,
                           random_state=1)
rf.fit(X_train, y_train)
```

```python
rmese = np.sqrt(((y_val - rf.predict(X_val)) ** 2).mean())
```

```python
print(rmese)
```

```python
# Question 3. Train random forest regressors with different numbers of trees: 10, 50, 100, and 200. For each model, calculate the RMSE on the validation dataset.
for n in [10, 50, 100, 200]:
    rf = RandomForestRegressor(n_estimators=n,
                               n_jobs=-1,
                               random_state=1)
    rf.fit(X_train, y_train)
    rmese = np.sqrt(((y_val - rf.predict(X_val)) ** 2).mean())
    print(n, round(rmese,3))
```

```python
# Question 4. Let's select the best max_depth
for max_depth in [10, 15, 20, 25]:
    for n in [10, 50, 100, 200]:
        rf = RandomForestRegressor(n_estimators=n,
                                   max_depth=max_depth,
                                   n_jobs=-1,
                                   random_state=1)
        rf.fit(X_train, y_train)
        rmese = np.sqrt(((y_val - rf.predict(X_val)) ** 2).mean())
        print(max_depth, n, round(rmese,3))
```

```python
# Question 5. What's the most important feature (among these 4)?
rf = RandomForestRegressor(n_estimators=10,
                           max_depth=20, 
                           n_jobs=-1,
                           random_state=1)
rf.fit(X_train, y_train)
print(rf.feature_importances_)
```

```python
feature_importance = pd.Series(
    rf.feature_importances_,
    index=dv.get_feature_names_out()
)

feature_importance.sort_values(ascending=False)
```

```python
# Question 6. Which eta leads to the best RMSE score on the validation dataset?
import xgboost as xgb
```

```python
features = list(dv.get_feature_names_out())
dtrain = xgb.DMatrix(X_train, label=y_train, feature_names=features)
dval = xgb.DMatrix(X_val, label=y_val, feature_names=features)
```

```python
xgb_params = {
    'eta': 0.3, 
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}
```

```python
model = xgb.train(xgb_params, dtrain, num_boost_round=100)
```

```python
y_pred = model.predict(dval)
rmse = np.sqrt(((y_val - y_pred) ** 2).mean())
print(f"RMSE: {rmse}")
```

```python
xgb_params = {
    'eta': 0.1, 
    'max_depth': 6,
    'min_child_weight': 1,
    
    'objective': 'reg:squarederror',
    'nthread': 8,
    
    'seed': 1,
    'verbosity': 1,
}
model = xgb.train(xgb_params, dtrain, num_boost_round=100)

y_pred = model.predict(dval)
rmse = np.sqrt(((y_val - y_pred) ** 2).mean())
print(f"RMSE: {rmse}")
```
